# 01 – Exploratory Data Analysis (Starter)

This notebook helps you quickly load a video-game sales dataset and sanity-check a few columns.
Place your CSVs under `data/raw/` before running.

```python
from pathlib import Path
import pandas as pd

raw_dir = Path('../data/raw')
candidates = [
    'vgchartz-2024.csv',
    'vgsales.csv',
    'Video_Games_Sales_as_at_22_Dec_2016.csv',
]
for name in candidates:
    path = raw_dir/name
    if path.exists():
        print('Loading', name)
        df = pd.read_csv(path)
        display(df.head(10))
        print('Shape:', df.shape)
        break
else:
    print('No CSVs found in ../data/raw — add your files first.')
```
